```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
df = pd.read_excel('data/Online Retail.xlsx')
total_transactions = df.groupby('CustomerID')['InvoiceNo'].nunique()

# 2. Zliczamy TYLKO anulowane transakcje (te z 'C')
canceled_transactions = df[df['InvoiceNo'].str.contains('C', na=False)].groupby('CustomerID')['InvoiceNo'].nunique()

# 3. Łączymy to do tabeli i liczymy wskaźnik
cancellation_df = pd.DataFrame({
    'Total_Orders': total_transactions,
    'Canceled_Orders': canceled_transactions
}).fillna(0) # Jeśli ktoś nie anulował nic, ma NaN, zmieniamy na 0

cancellation_df['Cancellation_Rate'] = cancellation_df['Canceled_Orders'] / cancellation_df['Total_Orders']
```
